In LangChain, instead of sending one giant text string to an AI model, conversations are structured as a list of Message objects. Every message has a specific role so the model knows who said what.

There are 4 core message types:

SystemMessage (The Rulebook / Persona):

What it is: Hidden instructions sent at the very beginning of the conversation to tell the AI how to behave, what role to play, or what rules to follow.

Example: "You are a helpful coding assistant. Always answer in 2 short sentences."

HumanMessage (The User's Input):

What it is: Whatever the user types, asks, or uploads (text, images, audio, or files).

Example: "How do I create a REST API?"

AIMessage (The Model's Reply):

What it is: The response generated by the AI model. Along with the text reply, it also stores metadata like token usage (usage_metadata) and any tool_calls the AI wants to run.

Bonus trick: You can also create an AIMessage manually in code and insert it into the message list to simulate past conversation history.

ToolMessage (The Tool's Result):

What it is: After an AIMessage asks to run a tool (like get_weather("Boston")), your Python code runs that function and sends the raw output back to the AI wrapped inside a ToolMessage.

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain.messages import SystemMessage, HumanMessage, AIMessage

load_dotenv(override=True)

model = init_chat_model("google_genai:gemini-3.5-flash-lite")

# Build a structured conversation history using message objects
messages = [
    SystemMessage("You are a helpful assistant that speaks simply."),
    HumanMessage("Can you help me with math?"),
    AIMessage("I'd be happy to help you with that!"),  # Simulated previous AI reply
    HumanMessage("Great! What is 15 * 4?")
]

response = model.invoke(messages)

print("Reply:", response.text)
print("Message Type:", type(response))
print("Token Usage:", response.usage_metadata)
```
