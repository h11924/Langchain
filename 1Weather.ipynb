```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv(override=True)

model = init_chat_model(
    "google_genai:gemini-3.5-flash-lite",
    timeout=15,
    max_retries=2,
)
response = model.invoke("Say hello in one short sentence.")
print(response.content)
```

making a weather basic agent


```python
from dotenv import load_dotenv
from langchain.agents import create_agent

load_dotenv(override=True)

# 1. Define a tool (ordinary Python function with type hints & docstring)
def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

# 2. Create the agent with your Gemini model, tools, and system prompt
agent = create_agent(
    model="google_genai:gemini-3.5-flash-lite",
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)

# 3. Run the agent by passing a user message
response = agent.invoke(
    {"messages": [{"role": "user", "content": "What is the weather in Hyderabad?"}]}
)

# 4. Print the full message trace and the final reply
print(response)
print("\nFinal Answer:", response["messages"][-1].text)
```
